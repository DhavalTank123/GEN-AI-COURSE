```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

prompt = [
    {
        "role": "system",
        "content": "You are A python developer "
    },
    {
        "role": "user",
        "content": "What is tree data structure in python? Explain with example."
    }
]
res = llm.invoke(prompt)
print(res.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",max_retries=5)

prompts = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "You are A Translator An translate input to {language} language" 
    },
    {
        "role": "user",
        "content": "{queary}"
    }
])
final_pormpts = prompts.format_prompt(language="Gujrati", queary="I am a devloper")
res = llm.invoke(final_pormpts)
print(res.content)

 
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",max_retries=5)
out = StrOutputParser()


def Transform_case(res:str):
    return res.upper()

prompts = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "You are A Translator An translate input to {language} language" 
    },
    {
        "role": "user",
        "content": "{queary}"
    }
])

# res = llm.invoke(prompts.invoke({
#     "language": "Gujrati",
#     "queary": "I am a devloper"
# }))
# print(res.content)

 
```

```python
chains = prompts | llm | out | Transform_case
res = chains.invoke({
    "language": "Gujrati",
    "queary": "I am a devloper"
})
print(res)
```
