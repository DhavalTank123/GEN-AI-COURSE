```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash",max_retries=5)

text = "User name is example and its email is example@gmail.com  and his age is 23."
res=llm.invoke(f"give me only Extracted the name,email and phone number from the following text: {text}")
res.text
```

```python
from pydantic import BaseModel,Field

class UserInfo(BaseModel):
    name: str = Field(description="The name of the user")
    email: str = Field(description="The email address of the user")
    age: int = Field(description="The age of the user")
    # phone_number: str = Field(description="The phone number of the user")
    
structured_llm = llm.with_structured_output(UserInfo)
structured_res = structured_llm.invoke(f"give me only Extracted the name,email, age  from the following text: {text}")
print(structured_res)
structured_res.model_dump()
print(structured_res.name)
print(structured_res.email)
print(structured_res.age)
# print(structured_res.phone_number)
```
