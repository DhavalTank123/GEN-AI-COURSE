```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",max_retries=5)


prompts = [
    {
        "role": "user",
        "content": "I am a Dhaval"
    },
    {
        "role":"ai",
        "content": "Hello Dhaval! It's nice to meet you. How can I help you today?"
    }
    ,{
        "role": "user",
        "content": "who am i?"
    }
] 
res = llm.invoke(prompts)
res.text
```

History

```python
history = []
while True:
    query = input("User: ")
    if query.lower() == "exit":
        break
    print(f"User: {query}")
    history.append({"role": "user", "content": query})
    
    res = llm.invoke(history)
    
    history.append({"role": "ai", "content": res.text})
    
    print(f"Ai: {res.text}")
```
