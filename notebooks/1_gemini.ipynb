```python
from IPython.display import Markdown, display
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
prompts = [
    ("system","You are a python developer."),#(type,) "system prompt","ai prompt","user prompt "
    ("user","Write a dsa a program for all sorting algorithms."),
]
res = llm.invoke(prompts)

# Extract clean text from res.content
if isinstance(res.content, list):
    # Extract text from the content dictionaries
    text_content = "".join([block["text"] for block in res.content if block.get("type") == "text"])
else:
    text_content = res.content

# Display as formatted Markdown in Jupyter Notebook
display(Markdown(text_content))
```
