```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("../Data/Dhaval_Tank_resume.pdf")
text=loader.load()

oneDoc = text[0]

splitter =  RecursiveCharacterTextSplitter(chunk_size = 50,chunk_overlap = 3  )
chunk = splitter.split_text(oneDoc.page_content)
chunk
```
